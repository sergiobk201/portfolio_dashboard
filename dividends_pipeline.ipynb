```python
import pandas as pd
```

```python
file = pd.read_excel('dividends.xlsx', skiprows=13)
file
```

```python
file = file.iloc[:, [1,2,3,5,6]]
file
```

```python
file = file.rename(columns={'Valor (R$)': 'Movement'})
file
```

```python
file.columns = ['movement_date', 'clearning_date','description','movement','balance']
file
```

```python
import datetime as dt

first_null = file['movement_date'].isna().argmax()
file = file.iloc[:first_null
                 ]
file
```

```python
file['movement_date'] = pd.to_datetime(file['movement_date'])
file['clearning_date'] = pd.to_datetime(file['clearning_date'])
file['movement_date'] = file['movement_date'].dt.date
file['clearning_date'] = file['clearning_date'].dt.date
file
```

```python
third_string = file['description'][2]
third_string.split()[0]
file['income_type'] = file['description'].str.split().str[0]
file
```

```python
word = file['description'][2].split()[-3]
word
```

```python
filtered = file[file['income_type'].isin(['JUROS','DIVIDENDOS'])]
filtered['ticker'] = filtered['description'].str.split().str[-3]
filtered['quantity'] = filtered['description'].str.split().str[-1]
filtered
```

```python
filtered['quantity'] = filtered['quantity'].astype('int32')
filtered['movement'] = filtered['movement'].astype('float64')
filtered['individual_movement'] = (filtered['movement'] / filtered['quantity']).round(2)
filtered
filtered
```
